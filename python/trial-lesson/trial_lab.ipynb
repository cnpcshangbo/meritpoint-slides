## Step 2. Give numbers names

A success rate is **goals ÷ attempts × 100%**. The denominator is the number of attempts, not the distance.

What do you predict this code will print?

```python
goals = 3
attempts = 5
rate = goals / attempts * 100
print(rate)
```

## Step 3. Let a list remember each attempt

`shots` holds the original five outcomes at 6 m. For this 0/1 list, `sum` counts goals and `len` counts attempts.

The mean (`sum / len`) is the success rate as a decimal; multiplying by 100 gives the percentage. Count the goals yourself before running the cell.

```python
shots = [1, 0, 1, 1, 0]
print(sum(shots))
print(len(shots))
print(sum(shots) / len(shots) * 100)
```

## Step 4. Compare three distances

Keep the original `shots` list unchanged. The accompanying `shots.csv` contains these same 15 invented attempts.

`goals_by_distance` is calculated from the lists. It records the **number of goals out of five attempts** for 3 m, 6 m, and 9 m. Its average, maximum, and minimum are goal counts, not percentages.

Predict the goal counts and success rates before running.

```python
distances_m = [3, 6, 9]
near = [1, 1, 1, 1, 0]
mid = shots.copy()
far = [0, 1, 0, 0, 1]

goals_by_distance = [sum(near), sum(mid), sum(far)]
average_goals = sum(goals_by_distance) / len(goals_by_distance)
print("Goals out of 5:", goals_by_distance)
print("Average goals:", average_goals)
print("Maximum goals:", max(goals_by_distance))
print("Minimum goals:", min(goals_by_distance))

rates = [
    sum(near) / len(near) * 100,
    sum(mid) / len(mid) * 100,
    sum(far) / len(far) * 100,
]
print("Distances (m):", distances_m)
print("Success rates (%):", rates)
```

## Step 5. Make the chart

Read the axis labels and the full 0–100% scale. Each bar represents five invented attempts at one distance.

The core commands are `plt.bar`, `plt.ylabel`, `plt.ylim`, and `plt.show`. The provided title, colors, labels, and save command make the chart easier to read and share; you can leave that scaffolding as it is.

The image is saved in the working folder as `success_by_distance.png`.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4.5))
bars = plt.bar(['3 m', '6 m', '9 m'], rates,
               color=['#147D78', '#E98664', '#89B6B6'])
plt.ylabel('Success rate (%)')
plt.ylim(0, 100)

# Provided chart scaffolding
plt.xlabel('Shot distance')
plt.title('Invented practice data: 5 attempts per distance')
plt.yticks([0, 20, 40, 60, 80, 100])
plt.gca().bar_label(bars, labels=[f'{value:.0f}%' for value in rates], padding=5)
plt.gca().set_axisbelow(True)
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.savefig('success_by_distance.png', dpi=180)
plt.show()
```
